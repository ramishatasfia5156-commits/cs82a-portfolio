## Part 1: Using a `for` Loop

```python
def multiply_for(num1, num2):
    """Multiply num1 by num2 using a for loop (repeated addition)."""
    result = 0
    for i in range(num2):
        result += num1
    return result

# Test calls
print("--- for loop version ---")
print(f"7 x 6  = {multiply_for(7, 6)}")
print(f"3 x 8  = {multiply_for(3, 8)}")
print(f"12 x 5 = {multiply_for(12, 5)}")
print(f"1 x 9  = {multiply_for(1, 9)}")
print(f"0 x 4  = {multiply_for(0, 4)}")
```

## Part 2: Using a `while` Loop

```python
def multiply_while(num1, num2):
    """Multiply num1 by num2 using a while loop (repeated addition)."""
    result = 0
    counter = 0
    while counter < num2:
        result += num1
        counter += 1
    return result

# Test calls
print("--- while loop version ---")
print(f"7 x 6  = {multiply_while(7, 6)}")
print(f"3 x 8  = {multiply_while(3, 8)}")
print(f"12 x 5 = {multiply_while(12, 5)}")
print(f"1 x 9  = {multiply_while(1, 9)}")
print(f"0 x 4  = {multiply_while(0, 4)}")
```

## How the `while` Loop Terminates

The while loop ends due to the counter variable: This begins as 0 and increases by +1 with each cycle (counter+=1). The conditions for the while loop are that counter must be less than num2. Therefore when counter = num2 the while loop condition becomes False. Because Counter will ALWAYS increment by EXACTLY 1 during each cycle, it WILL eventually equal num2; thereby ending the loop within a finite amount of cycles.

## Part 3: Verification : Both Functions Agree with `*`

```python
test_cases = [(7, 6), (3, 8), (12, 5), (1, 9), (0, 4)]

print(f"{'num1':>5} {'num2':>5} {'for result':>12} {'while result':>14} {'* result':>10} {'match?':>8}")
print("-" * 60)

all_correct = True
for num1, num2 in test_cases:
    for_result   = multiply_for(num1, num2)
    while_result = multiply_while(num1, num2)
    star_result  = num1 * num2
    match = (for_result == while_result == star_result)
    if not match:
        all_correct = False
    print(f"{num1:>5} {num2:>5} {for_result:>12} {while_result:>14} {star_result:>10} {'YES' if match else 'NO':>8}")

print("-" * 60)
print(f"All results correct: {all_correct}")
```
