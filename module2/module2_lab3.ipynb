## Temperature Cutoff Rules

All temperatures are in degrees Celsius (°C), recorded over a week.

| Category | Rule                              |
|----------|-----------------------------------|
| Cold     | temperature < 15°C                |
| Warm     | 15°C <= temperature <= 25°C       |
| Hot      | temperature > 25°C                |

**Boundary rule:** 15°C is *Warm* (inclusive lower bound); 25°C is *Warm* (inclusive upper bound); anything strictly above 25°C is *Hot*.

```python
# Problem 4.4 — Temperature Categories
# Categorizes each day's temperature as Cold, Warm, or Hot.

Temperatures = [10, 16, 23, 28, 14, 22, 30]

print(f"{'Temperature':>13}  {'Category'}")
print("-" * 26)

for temp in Temperatures:
    if temp < 15:
        category = "Cold"
    elif temp <= 25:
        category = "Warm"
    else:
        category = "Hot"

    print(f"{temp:>11}°C  {category}")
```

## Boundary Case Verification

Testing the exact boundary values (14, 15, 25, 26) to confirm the `elif` chain matches the stated cutoff rules.

```python
# Boundary spot-check
boundary_tests = [
    (14, "Cold"),
    (15, "Warm"),
    (25, "Warm"),
    (26, "Hot"),
]

print("Boundary verification:")
print(f"{'Temp':>6}  {'Expected':>8}  {'Got':>8}  {'Pass?'}")
print("-" * 36)

all_pass = True
for temp, expected in boundary_tests:
    if temp < 15:
        got = "Cold"
    elif temp <= 25:
        got = "Warm"
    else:
        got = "Hot"

    result = "PASS" if got == expected else "FAIL"
    if got != expected:
        all_pass = False
    print(f"{temp:>5}°C  {expected:>8}  {got:>8}  {result}")

print()
print("All boundary cases passed!" if all_pass else "One or more boundary cases FAILED.")
```
